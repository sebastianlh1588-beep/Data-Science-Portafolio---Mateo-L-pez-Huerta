# Eliminacion de los datos definidos como "No determinados" de la Variable Objetivo "Indice"
A continuacion eliminaramos los datos de la variable "Índice" que esten definidos como "no determinado", ya que no aportan nada a nuestro analisis, no existe evidencia de que tengan un mayor riesgo.

```python
#Eliminar del dataset df, de la variable objetivo "Indice" el valor como "No determinado "
df_modelo = df[df["Índice"] != "No determinado"].copy()
print(df_modelo.info())
```

```python
#Convertimos la variable"Índice" de (object) a numerica decimal (float)
df_modelo["Índice"] = pd.to_numeric(
df_modelo["Índice"],
errors="coerce"
)
df_modelo.reset_index(drop=True, inplace=True) # --- Reiniciamos el Indice del DataFrame df_modelo
print(df_modelo.info())
```

# Interpretación
Ahora hemos creado un nuevo dataframe llamado df_modelo, en el cual, hemos eliminado los registros correspondientes a "No determinado" de la variable "Indice", con ello, tenemos 89030 registros en total, ademas hemos cambiado el tipo de variable de (object) a (float), y por ultimo se ha reseteado el indice de nuestro dataframe

# Normalizacion de Variables
A continuacion vamos a normalizar nuestras variables, monto de adjudicación e indice

```python
# Normalizar monto
df_modelo["Monto_Norm"] = scaler.fit_transform(
df_modelo[[" Monto de Adjudicación"]]
)

# Normalizar índice
df_modelo["Indice_Norm"] = scaler.fit_transform(
df_modelo[["Índice"]]
)
```

```python
print(df_modelo.head())
```

#  Interpretación
¿Porque es importante normalizar estas variables? Esto nos permite, que las variables esten en escalas iguales para compararlas. Como se evidencia se han creado dos variables dentro de nuestro dataframe "Monto_Norm" e "Indice_Norm", estan normalizadas en una escala del 0 al 1, lo cual las hace comparables

# Creamos Dataframe solo con los procesos de contratacion de ETAPA EP
A continuacion se va a crear otro dataframe solo con los procesos de ETAPA EP, ya que el resto de procesos no las llevo a cabo la empresa y por lo tanto, no tiene sentido analizarlos.

```python
df_etapa = df_modelo[
df_modelo["Entidad"].str.upper() ==
"EMPRESA PÚBLICA MUNICIPAL DE TELECOMUNICACIONES, AGUA POTABLE, ALCANTARILLADO Y SANEAMIENTO DE CUENCA ETAPA EP"
].copy()
```

```python
#Verificamos la data de etapa
df_etapa.info()
```

# Interpretación
Ahora tenemos un nuevo Dataframe con el nombre "df_etapa", tiene 928 registros y 19 columnas, con 11 variables (object) y 8 variables con numeros decimales (float)

# Construir el Score de Priorizacion de Procesos de Contratación
Conocemos que el indice corresponde al promedio de: competencia, trazabilidad, transparencia, Temporabilidad, por lo tanto nuestro analisis sera en base solo al Índice ya que incluye el resto de indices nombrados, ademas, sabemos que si se acercan a uno, mejor sera el score de calificación, en ese caso vamos  invertir para que nos diga, si está mas bajo, existe mas riesgo de auditoria. Luego se establecera los siguientes criterios y pesos: Índice 70% Monto 30%.

```python
#Invertimos
df_etapa["Indice_Riesgo"] = 1 - df_etapa["Indice_Norm"]
df_etapa["Monto_Riesgo"] = 1 - df_modelo["Monto_Norm"]

#Calculamos el Score de Auditoría
df_etapa["Score_Auditoria"] = (
0.70 * df_etapa["Indice_Riesgo"]
+ 0.30 * df_etapa["Monto_Riesgo"])
```

# Ranking de procesos de auditoria
A continuacion se establecerá el ranking de procesos a auditar para el PAC 2027 de ETAPA EP

```python
ranking_auditoria = (
df_etapa
.sort_values(
by="Score_Auditoria",
ascending=False
)
)
```

# TOP 20 Procesos de Contratacion
Establaceremos y Observaremos, cuales son los 20 procesos mas riesgosos que se auditaran en base al score (criterios dados)

```python
ranking_auditoria[
[
"Código del proceso",
"Objeto",
" Monto de Adjudicación",
"Índice",
"Score_Auditoria"
]
].head(20)
```

# Clasificación del PAC
Ahora vamos a clasificar y asignar el nivel de prioridad de los procesos de contratacion del dataframe de ETAPA EP (df_etapa)

```python
q80 = df_etapa["Score_Auditoria"].quantile(0.80)
q95 = df_etapa["Score_Auditoria"].quantile(0.95)

df_etapa["Nivel_Prioridad"] = np.where(
df_etapa["Score_Auditoria"] >= q95,
"CRITICA",
np.where(
df_etapa["Score_Auditoria"] >= q80,
"ALTA",
"NORMAL"
)
)
```

# Interpretación
Al clasificarlos por q80 y q95 estamos haciendo los siguiente:

q95 = valor por encima del cual se encuentra el 5% de los procesos con mayor score, el cual se considera procesos con riesgo critico para realizar una auditoria.

q80 = valor por encima del cual se encuentra el 20% de los procesos con mayor score, el cual se considera procesos con riesgo alto para realizar una auditoria.

Los procesos de contraccion que esten por debajo de este 20% se consideran con riesgo normal, para realizar una auditoría


# Seleccionar procesos de Auditoria
Finalmente seleccionamos los procesos que segun su score de calificacion, se priorizaran para una revisión del PAC

```python
procesos_pac = df_etapa[
df_etapa["Nivel_Prioridad"].isin(
["CRITICA", "ALTA", "NORMAL"]
)
]
procesos_pac.head()
```
